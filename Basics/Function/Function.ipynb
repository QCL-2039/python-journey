```python
def remove_duplicate(yourlist):
    yourlist = yourlist[:]  # makes a shallow copy
    myset = set(yourlist)
    print("After removing the duplicate:", myset)
    yourlist.sort()
    print("Sorted copy:", yourlist)

mylist = [1,2,3,4,3,2,2,1,2,0]
remove_duplicate(mylist)
print("Original list:", mylist)
```

```python
def build_profile(first, last, **user_info):
    """Build a dictionary containing everything we know about a user."""
    profile = {}
    profile['first_name'] = first
    profile['last_name'] = last
    for key, value in user_info.items():
        profile[key] = value
    return profile
user_profile = build_profile('albert', 'einstein',
                             location='princeton',
                             field='physics')
print(user_profile)
```
